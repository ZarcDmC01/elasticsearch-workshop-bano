Mots clef :   

Big Data: (Lucas) Immenses volumes de données complexes qu'une base de données traditionnelle ne peut pas traiter.  

Il se caractérise principalement par les 3 V (il y en a d’autres, jusqu’à 7 au total - cf. pdf): 

Volume : une quantité massive de données générées en continu (teraoctets, pétaoctets). 

Vitesse : la nécessité de traiter et d'analyser ces flux en temps réel ou quasi réel. 

Variété : la diversité des formats (textes, images, vidéos, logs, données capteurs IoT, structurées ou non). 

Stack elastic: (Lucas) Ensemble de logiciels open-source (ou avec abonnements payants) développés par Elastic qui fonctionnent ensemble pour collecter, enrichir, stocker, analyser et visualiser des données en temps réel. On distingue 3 couches: la couche d’ingestion des données (Logstash + Beats), la couche de stockage et analyse (Elasticsearch) et la couche de visualisation et gestion (Kibana). 

ElasticSearch: (Lucas) Brique de stockage, recherche et analyse de la stack globale d’Elastic.  

Elastic Cloud: (Claude) Offre Elasticsearch managée par Elastic (SaaS), qui evite d'installer et de maintenir soi-meme les clusters. 

Kibana: (Claude) Interface web de la stack Elastic. Permet d'explorer les donnees (Discover), de creer des visualisations et dashboards, d'utiliser Dev Tools (console de requetes REST) et de gerer les index/mappings. 

Beats: (Claude) Famille d'agents legers ecrits en Go, installes au plus pres de la source de donnees (serveur, poste, conteneur) pour collecter des logs, metriques ou evenements et les envoyer vers Logstash ou directement Elasticsearch. 

Saas: (Software as a Service)  

Logstash: (Claude) Pipeline de traitement de donnees cote serveur : recupere des donnees depuis de multiples sources (input), les transforme/enrichit (filter), puis les envoie vers une ou plusieurs destinations (output), typiquement Elasticsearch. 

Apache Lucene: (Claude) Bibliotheque Java open source d'indexation et de recherche plein texte. C'est le moteur bas niveau sur lequel Elasticsearch est construit (index inverse, scoring BM25, etc.). 

BANO: (Claude) Base Adresses Nationale Ouverte, initiative d'OpenStreetMap France. Base de donnees libre regroupant les points d'adresse (numero, rue, code postal, commune, coordonnees GPS) a l'echelle de la France, utilisee comme dataset pour ce brief. 

Ingest pipeline: (Claude) Sequence de traitements (processors) appliquee aux documents cote Elasticsearch au moment de l'indexation (avant stockage) : renommage de champs, parsing, geocodage, enrichissement via un index de reference, etc. 

NoSQL: (Claude) Categorie de bases de donnees ne reposant pas sur le modele relationnel/SQL classique. Elasticsearch est un moteur NoSQL oriente documents JSON, optimise pour la recherche plutot que pour les transactions. 

Cluster: A group of nodes that work together to operate Elasticsearch. 

Node: A Java process that runs the Elasticsearch software. 

Index: (Claude) Equivalent d'une "table" dans Elasticsearch : une collection de documents JSON partageant un mapping commun. Un index est physiquement decoupe en un ou plusieurs shards. 

Shards: (Lucas) Un Shard est une unité de stockage de base autonome. C'est une portion physique d'un index Elasticsearch. Chaque shard peut ainsi être stocké sur un serveur différent, cela permet la scalabilité horizontale. 

Segment: (Claude) Sous-unite d'un shard. Lucene stocke les donnees d'un shard sous forme de plusieurs segments immuables (fichiers d'index), periodiquement fusionnes (merge) pour optimiser les performances de recherche. 

Replica: (Lucas) Copie d’un shard primaire qui va être stockée sur un noeud (node) différent. Elle permet la tolérance aux pannes.  

Mapping: (Claude) Schema d'un index : definit pour chaque champ son type de donnees (text, keyword, geo_point, dense_vector...) et comment il doit etre indexe/analyse. Peut etre defini explicitement ou deduit automatiquement (dynamic mapping). 

cURL: (Lucas) ClientURL. Outil en ligne de commande permettant de récupérer ou transférer des données via divers protocoles réseaux (HTTP, HTTPS, FTP, SMTP etc). Il repose sur la librairie libcurl et est très utilisé, notamment pour interagir avec des API REST, télécharger des fichiers ou tester des serveurs web. 

Hive: (Claude) Outil de l'ecosysteme Hadoop permettant d'interroger en SQL (HiveQL) de tres gros volumes de donnees stockees sur un systeme de fichiers distribue (HDFS). Contrairement a Elasticsearch, Hive est oriente traitement batch, pas recherche temps reel. 

Container: (Claude) Unite logicielle isolee (processus + dependances + filesystem) partageant le noyau de l'OS hote, executee par un moteur comme Docker. Utilise ici pour lancer Elasticsearch et Kibana sans les installer directement sur la machine. 

kubernetes: (Claude) Orchestrateur de conteneurs permettant de deployer, scaler et gerer automatiquement des clusters de conteneurs (par exemple des noeuds Elasticsearch) en production. Non utilise dans ce brief (usage local via Docker seul). 

Ves: (Claude) [terme non identifie / probable coquille dans les notes originales - a clarifier avec le groupe]. 

Docker: (Claude) Plateforme de conteneurisation utilisee pour packager et lancer Elasticsearch et Kibana de facon isolee et reproductible en local, via le script officiel start-local (docker compose). 

ESQL (elastic search query language): (Claude) Langage de requete de type pipeline (proche de SQL/piping shell) integre a Elasticsearch et exploitable directement dans Kibana Discover, permettant de filtrer, transformer et agreger les donnees sans ecrire de JSON Query DSL. 

JVM Heap: (Claude) Memoire allouee a la machine virtuelle Java (JVM) qui fait tourner Elasticsearch. Un dimensionnement correct (generalement max 50% de la RAM disponible, sans depasser ~32Go) est critique pour la performance et la stabilite du cluster. 

OS Cache: (Claude) Cache disque gere par le systeme d'exploitation (page cache), utilise par Lucene pour garder en memoire les fichiers d'index frequemment lus, en complement du JVM Heap. 

Thread Pool: (Claude) Groupe de threads dedies a un type d'operation (recherche, indexation, bulk...) sur un noeud Elasticsearch, avec une file d'attente associee pour lisser la charge. 

Thread Queues: (Claude) File d'attente associee a un thread pool : les requetes en exces par rapport aux threads disponibles y patientent avant d'etre rejetees si la queue est pleine (protection contre la surcharge du noeud). 

Cross Cluster search (CCS): (Claude) Fonctionnalite permettant d'executer une seule requete de recherche sur plusieurs clusters Elasticsearch distincts (par exemple repartis geographiquement), sans avoir a fusionner les donnees au prealable. 

 

 

 

 

Hypothèses :    

Jesse : Elastic search permet de traiter de la véracité des données. 

(Claude) A NUANCER. Elasticsearch ne verifie pas la 'veracite' (exactitude factuelle) d'une donnee au sens propre - il ne fait pas de fact-checking. En revanche, il aide a la qualite/coherence des donnees via le mapping (typage strict), les contraintes d'indexation, et la possibilite de detecter des doublons ou incoherences par des requetes d'agregation. La 'veracite' est le V du Big Data le moins directement adresse par Elasticsearch lui-meme : elle depend surtout de la qualite de la source (ici BANO, alimentee par la communaute OpenStreetMap). 

 

Comparaison 

Comparer elastic search et kibana 

(Claude) 

| Critere | Elasticsearch | Kibana | 
|---|---|---| 
| Role | Moteur de stockage, indexation et recherche/analyse des donnees | Interface de visualisation, exploration et administration |

Conclusion : les deux sont complementaires et non substituables - Elasticsearch est le moteur, Kibana est la vitrine.

## Réponses aux problématiques (Claude)

**1) En quoi Elasticsearch est plus adapté pour le Big Data ?**

Contrairement à un SGBD relationnel (comme le MsSQL utilisé par l'API existante), Elasticsearch est construit dès le départ pour la recherche et l'analyse de gros volumes de données semi-structurées :
- Indexation en **index inversé** (Lucene) : recherche textuelle en quasi temps réel même sur des millions de documents.
- **Scalabilité horizontale** native via les shards : un index se répartit sur plusieurs nœuds sans réécrire l'application.
- Pas de schéma rigide imposé à l'avance (mapping dynamique possible), utile pour des données hétérogènes (le "Variété" du Big Data).
- API REST/JSON, donc facilement scriptable et intégrable dans un pipeline de collecte automatisée (voir C1 du brief).

**2) Comment ingérer des données dans le stack Elastic ?**

Plusieurs méthodes, illustrées concrètement dans ce projet :
- Manuellement via l'API REST (`POST /index/_doc/id`, `_bulk`).
- Par script/client (Python `elasticsearch` + `helpers.bulk`, voir [index_bano.py](../index_bano.py)) : c'est la méthode utilisée pour charger les 203 672 adresses BANO du département 25.
- Via Logstash/Beats pour des flux continus (logs, capteurs) — non nécessaire ici car le dataset est un fichier CSV statique.

**3) Comment réaliser des requêtes simples et complexes avec Elasticsearch ?**

- Requêtes simples : `match` (recherche textuelle), `term`/`filter` (valeur exacte), testées via curl et Dev Tools.
- Requêtes complexes : `bool` (combinaison should/filter), `geo_distance` (recherche par rayon GPS) et agrégations `terms` (regroupement/statistiques), implémentées et testées dans [search_bano.py](../search_bano.py) — exemple : trouver les adresses les plus proches d'un point GPS, ou compter les adresses par commune.

**4) Comment utiliser le stack Elastic en local ?**

Via Docker, sans dépendre du Cloud Elastic : le script officiel `start-local` lance deux conteneurs (`es-local-dev`, `kibana-local-dev`) préconfigurés avec authentification basique et une clé API, accessibles sur `localhost:9200` (Elasticsearch) et `localhost:5601` (Kibana).

**5) Comment installer le stack Elastic avec Docker ?**

```bash
curl -fsSL https://elastic.co/start-local | sh
```

Ce script télécharge un `docker-compose` préconfiguré, crée le dossier `elastic-start-local/` (contenant le `.env` avec les identifiants générés), tire les images officielles `docker.elastic.co/elasticsearch` et `docker.elastic.co/kibana`, puis démarre les conteneurs. Résultat obtenu dans ce projet : Elasticsearch 9.5.3 + Kibana opérationnels en local en moins de 5 minutes.

**6) Comment utiliser l'API REST de la stack Elastic pour rechercher et indexer les données ?**

- Indexation : `PUT /<index>` (création + mapping), `POST /<index>/_doc/<id>` (un document), `PUT /<index>/_bulk` (NDJSON, plusieurs documents d'un coup — utilisé pour les 203k adresses).
- Recherche : `GET /<index>/_search` avec un corps JSON décrivant la requête (Query DSL) ou des agrégations.
- Authentification : `Authorization: ApiKey <clé>` ou Basic Auth (`-u elastic:<password>`), comme configuré dans le `.env` généré par start-local.


## Résultats pratiques obtenus sur ce brief (Claude)

- **Environnement** : Elasticsearch 9.5.3 + Kibana installés en local via Docker (`start-local`), accessibles sur `http://localhost:9200` et `http://localhost:5601`.
- **Dataset** : BANO département 25 (Doubs) téléchargé depuis `bano.openstreetmap.fr` — 203 672 adresses.
- **Indexation** : script [index_bano.py](../index_bano.py) — création d'un mapping dédié (champ `location` en `geo_point`, sous-champs `.keyword` pour les agrégations), puis chargement en masse via l'API `_bulk` (0 erreur).
- **Recherche** : script [search_bano.py](../search_bano.py) —
  - recherche par nom (numéro + rue + filtre commune) ;
  - recherche géospatiale (`geo_distance` + tri par distance), validée avec un point GPS réel du centre de Besançon (résultat le plus proche à 5 mètres).
- **Agrégation** : test `terms` sur `commune.keyword` → top communes du département : Besançon (19 674 adresses), Montbéliard (5 141), Pontarlier (4 725), Audincourt (4 309), Valentigney (3 321).
- **Alias** : création d'un alias `bano` pointant vers l'index `bano-25`, pour préparer une organisation multi-départements (un index par département + un alias global), comme recommandé par l'article Elastic *"Enriching Your Postal Addresses With the Elastic Stack - Part 2"* (Workshop 2).

### Non réalisé (pistes de poursuite)

- Pipeline **Logstash** d'enrichissement automatique (routage dynamique par département depuis le code postal, filtre `elasticsearch` pour enrichir un flux entrant en continu).
- **Analyzer français** personnalisé sur les champs `rue`/`commune` pour améliorer la tolérance aux fautes d'orthographe et accents (piste d'amélioration citée dans les critères de performance du brief).
- Dépôt des requêtes détaillées sur GitHub (deuxième livrable du brief).


## Conclusion (Claude)

Le brief avait pour objectif de démontrer qu'Elasticsearch constitue une alternative crédible à une API de recherche d'adresses postales reposant sur des requêtes SQL classiques. Les tests réalisés sur le dataset BANO du département 25 confirment cette hypothèse :

- L'indexation de 203 672 documents s'est faite en quelques secondes, sans erreur.
- Les recherches textuelles et géospatiales renvoient des résultats pertinents et rapides (< 20 ms), avec une tolérance native aux variations texte qu'une requête `LIKE` SQL n'offre pas.
- Les agrégations permettent des analyses instantanées (répartition par commune) impossibles à obtenir aussi simplement en SQL classique sans requêtes `GROUP BY` coûteuses sur de gros volumes.

Elasticsearch (couplé à Kibana pour l'exploration visuelle) répond donc bien au besoin exprimé par l'équipe de développement : une méthode alternative aux requêtes SQL, plus adaptée à la recherche textuelle floue et à l'analyse de gros volumes de données géolocalisées.
